Membuat Model Regresi Linier

```python
# 1. Import library
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression

# 2. Membaca dataset
df = pd.read_csv('toyota.csv')

# 3. Menentukan fitur dan target
features = ['year', 'mileage', 'tax', 'mpg', 'engineSize']

X = df[features]
y = df['price']

# 4. Membagi data training dan testing
X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.2,
    random_state=70
)

# 5. Membuat model Regresi Linier
lr = LinearRegression()

# 6. Melatih model
lr.fit(X_train, y_train)

# 7. Melakukan prediksi
pred = lr.predict(X_test)

print("Model Regresi Linier berhasil dibuat!")
print("Jumlah data training:", len(X_train))
print("Jumlah data testing:", len(X_test))

# 8. Menampilkan hasil prediksi
hasil = pd.DataFrame({
    'Harga Aktual': y_test.values,
    'Harga Prediksi': pred
})

display(hasil.head(10))
```

```python
score = lr.score(X_test, y_test)
print('akurasi model regresi linier = ', score)
```

Membuat Inputan Model Regresi Linear

```python
#Year=2019, mileage=5000, tax=145, mpg=30.2, engineSize=2
input_data = np.array([[2019,5000,145,30.2,2]])

prediction = lr.predict(input_data)
print('Estimasi harga mobil dalam EUR :', prediction)
```

Save Model

```python
# 1. Import library
import pandas as pd
import numpy as np
import pickle

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression

# 2. Membaca dataset
df = pd.read_csv('toyota.csv')

# 3. Menentukan fitur dan target
features = ['year', 'mileage', 'tax', 'mpg', 'engineSize']

X = df[features]
y = df['price']

# 4. Membagi data training dan testing
X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.2,
    random_state=70
)

# 5. Membuat model Regresi Linier
lr = LinearRegression()

# 6. Melatih model
lr.fit(X_train, y_train)

# 7. Menyimpan model
filename = 'estimasi_mobil.sav'

with open(filename, 'wb') as file:
    pickle.dump(lr, file)

print("Model berhasil disimpan!")
print("Nama file:", filename)
```
